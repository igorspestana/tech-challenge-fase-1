```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Carregar a imagem
imagem = cv2.imread('./content/perfil.jpeg')


# Exibir a imagem carregada
cv2.imshow('Imagem', imagem)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Converter a imagem de BGR para RGB
imagem_rgb = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)

# Exibir a imagem
plt.imshow(imagem_rgb)

# plt.axis('off') # Desabilita os eixos

plt.show()
```

```python
# Converter imagem para grayscale
imagem_gray = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)

# Exibir a imagem grayscale
plt.imshow(imagem_gray, cmap='gray')
plt.axis('off') # Desabilita os eixos

plt.show()
```

#### Aplicação de filtros para suavização de imagens

O que ajuda a reduzir o ruído e na melhora da detecção de bordas. (Fica meio blur)

```python
# pega a imagem e aplica o filtro nela

imagem_suavizada =  cv2.GaussianBlur(imagem, (15,15), 0)

# Converte para rgb
imagem_suavizada_rgb = cv2.cvtColor(imagem_suavizada, cv2.COLOR_BGR2RGB)

# Exibe a imagem suavizada
plt.imshow(imagem_suavizada_rgb)

plt.axis('off') # Desabilita os eixos

plt.show()
```

### Detecção de bordas com canny

Passo crucialem muitos algoritmos de visão computacional, principalmente de detecção de objetos.

```python
# pega a imagem de grayscale

# detecta bordas na imagem
bordas = cv2.Canny(imagem_gray, 100,200)

# Exibir a imagem com bordas detectadas
plt.imshow(bordas, cmap='gray')
plt.axis('off') # Desabilita os eixos
plt.show()
```

### Desenhar formas na imagem

É possível desenhar formas como retângulos para destacar áreas de interesse na imagem

```python
# Definir as coordenadas e a cor do retângulo
inicio = (50, 5)
fim = (220, 160)
cor = (255, 0, 0)
espessura = 2

# Desenhar o retângulo na cópia da imagem
imagem_com_retangulo = cv2.rectangle(imagem.copy(), inicio, fim, cor, espessura)

# Converter a imagem para RGB
imagem_com_retangulo_rgb = cv2.cvtColor(imagem_com_retangulo, cv2.COLOR_BGR2RGB)

# Exibir a imagem com retângulo
plt.imshow(imagem_com_retangulo_rgb)

plt.axis('off') # Desabilita os eixos

plt.show()
```

### Salvando a imagem processada

```python
caminho_salvar = './imagem_processada.jpg'
cv2.imwrite(caminho_salvar, imagem_com_retangulo)
```
